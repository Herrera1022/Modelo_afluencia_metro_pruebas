4. Al analizar el funcionamiento en horarios del sistema metro, identificamos que cada línea tiene horarios diferentes para cada día, por lo tanto, los valores nulos de cada línea por fuera de su horario de funcionamiento decidimos eliminarlos, ya que no son registros validos debido a que la línea no estaba operando.

Horarios lineas A - B - TA - H - J - K - M - O - P - 1 - 2

Lunes - Sabado: 4:30am - 11:00pm

Horarios lineas A - B - TA - O - 1 - 2

Domingos - Festivos: 5:00 am - 10pm

Horarios lineas K - H - J - M - P

Domingos - Festivos: 9:00 am - 10pm

Horarios linea L - 9:00am - 6:00pm (Fijo a esa fecha)


```python
# Tabla de horarios operativos por línea y tipo de día
# tipo_dia: 0 = lunes-sabado, 1 = domingo/festivo
horarios = {
    # linea: {tipo_dia: (hora_inicio, hora_fin)}
    "LÍNEA 1":  {0: (4, 23), 1: (5, 22)},
    "LÍNEA 2":  {0: (4, 23), 1: (5, 22)},
    "LÍNEA A":  {0: (4, 23), 1: (5, 22)},
    "LÍNEA B":  {0: (4, 23), 1: (5, 22)},
    "LÍNEA T-A":{0: (4, 23), 1: (5, 22)},
    "LÍNEA O":  {0: (4, 23), 1: (5, 22)},
    "LÍNEA H":  {0: (4, 23), 1: (9, 22)},
    "LÍNEA J":  {0: (4, 23), 1: (9, 22)},
    "LÍNEA K":  {0: (4, 23), 1: (9, 22)},
    "LÍNEA M":  {0: (4, 23), 1: (9, 22)},
    "LÍNEA P":  {0: (4, 23), 1: (9, 22)},
    "LÍNEA L":  {0: (9, 18), 1: (9, 17)},  # horario fijo todos los días
}

def es_fuera_de_horario(row):
    tipo_dia = 1 if (row["dia_semana"] == 6 or row["festivo"] == 1) else 0
    h_inicio, h_fin = horarios[row["linea"]][tipo_dia]
    return row["hora"] < h_inicio or row["hora"] > h_fin

# Mask: nulo Y fuera de horario
mask_eliminar = (
    df_long["pasajeros"].isna() &
    df_long.apply(es_fuera_de_horario, axis=1)
)

print(f"Filas antes:          {len(df_long)}")
print(f"Filas a eliminar:     {mask_eliminar.sum()}")

df_long = df_long[~mask_eliminar].reset_index(drop=True)

print(f"Filas después:        {len(df_long)}")
print(f"Nulos restantes:      {df_long['pasajeros'].isna().sum()}")
```

Ahora vamos a codificar las variables, como notamos las variables dia_semana, semana, festivo ya estan numericos y en rangos pequeños, por lo tanto vamos a decidir dejarlos así y no normalizar su escala. Para las variables hora y linea que son categoricas se aplicaron dos estrategias distintas según la naturaleza de cada variable. La variable hora se trató con Label Encoding (conversión directa a entero), ya que representa una variable ordinal con un orden cronológico natural — las horas tienen una progresión lógica que los modelos pueden aprovechar directamente. Por su parte, la variable linea se codificó con One-Hot Encoding, ya que es una variable nominal donde no existe ningún orden inherente entre las líneas del metro, por lo que asignarle valores enteros introduciría una jerarquía artificial que podría sesgar el modelo.

```python
#Hora
df_long["hora"] = df_long["hora"].astype(int)

# Hora como variable cíclica

df_long["hora_sin"] = np.sin(2 * np.pi * df_long["hora"] / 24)

df_long["hora_cos"] = np.cos(2 * np.pi * df_long["hora"] / 24)

# Semana como variable cíclica

df_long["semana_sin"] = np.sin(2 * np.pi * df_long["semana"] / 52)

df_long["semana_cos"] = np.cos(2 * np.pi * df_long["semana"] / 52)

#Linea
dummies = pd.get_dummies(df_long["linea"], prefix="linea")
df_long = pd.concat([df_long, dummies], axis=1)

# Verificación
print(df_long.shape)
print(df_long.dtypes)
print(df_long.head())
```

# Cronología

Debido a que la afluencia de pasajeros corresponde a un fenómeno temporal, las variables originales no eran suficientes para capturar la dependencia entre observaciones consecutivas. Con únicamente variables como semana, día de la semana, hora y línea, el modelo tendería a generar predicciones muy similares para intervalos de tiempo equivalentes, ignorando el comportamiento reciente del sistema.

Para incorporar información temporal al modelo, se agregaron variables históricas calculadas sobre la misma línea del sistema de transporte. Estas variables permiten introducir memoria temporal y capturar patrones de corto y mediano plazo en la demanda de pasajeros.

Las variables históricas seleccionadas fueron:

- La cantidad de pasajeros registrada en la hora inmediatamente anterior.
- El promedio de pasajeros de la misma línea durante el día inmediatamente anterior.
- La cantidad de pasajeros registrada en la misma hora de la semana anterior.

Con estas características el modelo puede considerar no solo el contexto temporal general, sino también el comportamiento reciente y los patrones periódicos de la serie de tiempo.

1. Ordenar el dataset por [linea -> fecha -> hora] para estructurarlo adecuadamente y obtener así los valores previos necesarios para las nuevas columnas

```python
# Ordenar cronológicamente por línea, fecha y hora
df_long = df_long.sort_values(["linea", "fecha", "hora"]).reset_index(drop=True)
```

```python
# Definir máscaras temporales 

mask_train      = df_long["fecha"] < "2024-10-01"
mask_validation = (df_long["fecha"] >= "2024-10-01") & (df_long["fecha"] < "2024-11-01")
mask_test       = df_long["fecha"] >= "2024-11-01"

# Imputar pasajeros ANTES de calcular variables históricas 
medianas_pasajeros = (
    df_long[mask_train]
    .groupby(["linea", "hora", "dia_semana"])["pasajeros"]
    .median()
)
medianas_pasajeros_fallback = (
    df_long[mask_train]
    .groupby(["linea", "hora"])["pasajeros"]
    .median()
)

def imputar_pasajeros(df):
    def fill(row):
        if pd.isna(row["pasajeros"]):
            key = (row["linea"], row["hora"], row["dia_semana"])
            if key in medianas_pasajeros.index:
                return medianas_pasajeros[key]
            return medianas_pasajeros_fallback.get((row["linea"], row["hora"]), 0)
        return row["pasajeros"]
    return df.apply(fill, axis=1)

df_long["pasajeros"] = imputar_pasajeros(df_long)
```

2. Generar la columna lag_1 que corresponde a la cantidad de pasajeros en cada linea, en la hora anterior.

```python
# Cantidad de pasajeros en la hora inmediatamente anterior
df_long["lag_1"] = (
    df_long
    .groupby(["linea", "fecha"])["pasajeros"]
    .shift(1)
)
```

Esta nueva columna tendrá valores faltantes en las primeras horas de operación de cada día, debido a que no existe una hora previa dentro del mismo día para calcular el rezago. Dado que en un futuro podría ser necesario predecir la cantidad de pasajeros durante la primera hora de operación para cualquier línea, estas muestras no deben ser eliminadas. Por esta razón, se creó una variable binaria indicadora de valores faltantes, con el objetivo de conservar la información asociada al inicio de operación de cada línea. Posteriormente, los valores faltantes fueron imputados con cero, ya que en presencia de la variable indicadora el modelo podrá diferenciar entre una ausencia de información histórica y un valor real de pasajeros igual a cero.

```python
#Crear indicador de valores faltantes
df_long["lag_1_missing"] = (
    df_long["lag_1"].isna().astype(int)
)

#Imputar valores faltantes
df_long["lag_1"] = df_long["lag_1"].fillna(0)
```

3. Generar la columna promedio_dia_anterior, correspondiente al promedio de pasajeros por hora del día anterior para cada línea. Este valor se repetirá en todas las muestras pertenecientes a la misma línea y fecha, debido a que representa un resumen del comportamiento operacional del día previo.

Adicionalmente, esta variable presentará valores faltantes en la primera fecha del dataset (2024-01-01), ya que no se cuenta con información del día anterior para calcular el promedio correspondiente. Debido a que estas muestras representan únicamente el primer día de operación registrado en el conjunto de datos y corresponden a una proporción muy pequeña del total de observaciones, dichas filas serán eliminadas.

```python
#Auxiliar para guardar el total de pasajeros y las horas de operación de una linea en un día
resumen_diario = (
    df_long
    .groupby(["linea", "fecha"])
    .agg(
        total_pasajeros=("pasajeros", "sum"),
        horas_operacion=("pasajeros", "count")
    )
    .reset_index()
)

#Calcular el promedio
resumen_diario["promedio_dia"] = (
    resumen_diario["total_pasajeros"] /
    resumen_diario["horas_operacion"]
).round().astype(int)

#Agregar una columna para el día anterior a cada línea
resumen_diario["promedio_dia_anterior"] = (
    resumen_diario
    .groupby("linea")["promedio_dia"]
    .shift(1)
)

#Concatenar a df_long
df_long = df_long.merge(
    resumen_diario[["linea", "fecha", "promedio_dia_anterior"]],
    on=["linea", "fecha"],
    how="left"
)
```

```python
#Eliminar muestras sin promedio del día anterior
df_long = df_long[
    df_long["promedio_dia_anterior"].notna()
]
```

```python
# Calcular medias de referencia SOLO con train
medias_train = (
    df_long[mask_train]
    .groupby(["linea", "hora"])["pasajeros"]
    .mean()
)

# Imputar train (enero - septiembre) con sus propias medias
df_long.loc[mask_train, "misma_hora_semana_anterior"] = (
    df_long.loc[mask_train, "misma_hora_semana_anterior"]
    .fillna(df_long.loc[mask_train].groupby(["linea", "hora"])["pasajeros"].transform("mean"))
)

# Imputar validation y test con medias del TRAIN (Validation: octubre, Test: noviembre-diciembre)
for mask in [mask_validation, mask_test]:
    idx = df_long[mask & df_long["misma_hora_semana_anterior"].isna()].index
    for i in idx:
        row = df_long.loc[i]
        df_long.loc[i, "misma_hora_semana_anterior"] = medias_train.get(
            (row["linea"], row["hora"]), 0
        )

#Comprobar cantidad de NaN por columna 

df_long.isna().sum()
```

```python
# Calcular medianas solo con train
medianas_pasajeros = (
    df_long[mask_train]
    .groupby(["linea", "hora", "dia_semana"])["pasajeros"]
    .median()
)
medianas_pasajeros_fallback = (
    df_long[mask_train]
    .groupby(["linea", "hora"])["pasajeros"]
    .median()
)

def imputar_pasajeros(df):
    def fill(row):
        if pd.isna(row["pasajeros"]):
            key = (row["linea"], row["hora"], row["dia_semana"])
            if key in medianas_pasajeros.index:
                return medianas_pasajeros[key]
            return medianas_pasajeros_fallback.get((row["linea"], row["hora"]), 0)
        return row["pasajeros"]
    return df.apply(fill, axis=1)

df_long["pasajeros"] = imputar_pasajeros(df_long)
```

Metodología de validación

Debido a la naturaleza temporal del problema, el dataset se dividirá respetando el orden cronológico de las observaciones. 

Train: enero - septiembre 2024 
Validation: octubre 2024 
Test: noviembre - diciembre 2024 

Esto permite entrenar el modelo usando únicamente información pasada para predecir periodos futuros, evitando data leakage.

```python
#Crear subconjuntos segun fecha 

train = df_long[mask_train] 
validation = df_long[mask_validation] 
test = df_long[mask_test] 

#Variables que no se usaran para entrenar 

columnas_excluir = [ "fecha", "linea", "pasajeros", "hora", "semana" ] 

#Separar X e y 

X_train = train.drop(columns=columnas_excluir) 
y_train = train["pasajeros"] 
X_validation = validation.drop(columns=columnas_excluir) 
y_validation = validation["pasajeros"] 
X_test = test.drop(columns=columnas_excluir) 
y_test = test["pasajeros"]
```

Estandarización de variables 

Debido a que las variables numéricas del dataset presentan escalas muy diferentes entre sí, se realizará un proceso de estandarización sobre las variables continuas. Este procedimiento permite que las variables tengan media cercana a 0 y desviación estándar cercana a 1, evitando que aquellas con magnitudes mayores dominen el proceso de entrenamiento del modelo.
